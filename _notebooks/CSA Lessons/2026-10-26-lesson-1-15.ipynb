# 1.15 String Manipulation

Original lesson: [1.15 String Manipulation](https://pages.opencodingsociety.com/csa/unit_01/1_15)

**Key idea:** a `String` is **immutable**. Methods like `substring`, `concat`, and `toUpperCase` never change the original string. They return a new `String` object. Wrapper classes (`Integer`, `Double`) let primitives act as objects, and autoboxing and unboxing convert between the two automatically.

## Popcorn Hack #1: Extend the Dog Class

I did both options. I added a second non-void method, `getAge()`, and a third, `getAgeInDogYears()`, and called them on two different `Dog` objects.

```java
public class Dog {
    public String Name;
    public int Age;

    public Dog(String name, int age) {
        Name = name;
        Age = age;
    }

    public String getName() {
        return Name;
    }

    // Additional non-void method: returns this dog's age
    public int getAge() {
        return Age;
    }

    // Non-void method that computes a value from this dog's own field
    public int getAgeInDogYears() {
        return Age * 7;
    }

    public static void main(String[] args) {
        Dog d = new Dog("Larry", 4);
        System.out.println(d.getName());

        Dog d2 = new Dog("Bella", 9);   // the same non-void methods on a second object
        System.out.println(d.getName() + " is " + d.getAge() + " (" + d.getAgeInDogYears() + " in dog years)");
        System.out.println(d2.getName() + " is " + d2.getAge() + " (" + d2.getAgeInDogYears() + " in dog years)");
    }
}

Dog.main(null);
```

## Popcorn Hack #2: Find a Substring's Position

`findPosition` wraps `indexOf`. The test prints the original `String` again afterward to prove that no method call changed it.

```java
public class SubstringFinder {
    // Returns the index where target first appears in full, or -1 if it is not there
    public static int findPosition(String full, String target) {
        return full.indexOf(target);
    }

    public static void main(String[] args) {
        String sentence = "hihihellothere";

        System.out.println("\"hello\" starts at index " + findPosition(sentence, "hello"));
        System.out.println("\"there\" starts at index " + findPosition(sentence, "there"));
        System.out.println("\"hi\" starts at index " + findPosition(sentence, "hi") + " (first occurrence)");
        System.out.println("\"bye\" starts at index " + findPosition(sentence, "bye") + " (not found)");

        // Check: pulling the substring back out at the found index gives the target
        int start = findPosition(sentence, "hello");
        System.out.println("substring check: " + sentence.substring(start, start + "hello".length()));

        // Immutability: substring/toUpperCase returned NEW strings; the original is unchanged
        String shouted = sentence.toUpperCase();
        System.out.println("toUpperCase result: " + shouted);
        System.out.println("original after all calls: " + sentence);
    }
}

SubstringFinder.main(null);
```

## Popcorn Hack #3: Math-Powered Number Guessing Game

The program plays the game itself. `Math.random()` picks a secret number from 1 to 100. A binary-search player then guesses, and `Math.abs()` measures how far off each guess is to give "hot/warm/cold" hints. `Math.pow()` scores the win, and the score doubles for every guess left unused. Binary search always wins within 7 guesses, because 2^7 = 128 ≥ 100. The secret number is random, so the output changes every time the cell runs.

```java
public class NumberGuessingGame {
    static final int MAX_NUMBER = 100;
    static final int MAX_GUESSES = 7;

    // Hint based on distance, computed with Math.abs
    public static String hint(int guess, int secret) {
        int distance = Math.abs(guess - secret);
        if (distance == 0) return "CORRECT!";
        String direction = guess < secret ? "higher" : "lower";
        if (distance <= 5) return "HOT - go " + direction;
        if (distance <= 20) return "warm - go " + direction;
        return "cold - go " + direction;
    }

    // Score doubles for each unused guess: 2^(guesses left) * 10
    public static int score(int guessesUsed) {
        return (int) (Math.pow(2, MAX_GUESSES - guessesUsed) * 10);
    }

    public static void main(String[] args) {
        int secret = (int) (Math.random() * MAX_NUMBER) + 1;   // 1..100
        int low = 1;
        int high = MAX_NUMBER;
        System.out.println("I'm thinking of a number from 1 to " + MAX_NUMBER + "...");

        for (int attempt = 1; attempt <= MAX_GUESSES; attempt++) {
            int guess = (low + high) / 2;   // binary-search player
            System.out.println("Guess " + attempt + ": " + guess + " -> " + hint(guess, secret));
            if (guess == secret) {
                System.out.println("Won in " + attempt + " guesses! Score: " + score(attempt));
                return;
            }
            if (guess < secret) {
                low = guess + 1;
            } else {
                high = guess - 1;
            }
        }
        System.out.println("Out of guesses! The number was " + secret);
    }
}

NumberGuessingGame.main(null);
```

## Extra: Immutability and Wrapper Classes (quiz review)

This cell practices the quiz topics. `+` builds a new `String`, and wrapper objects come from `valueOf` and autoboxing, not the deprecated `new Integer(...)`.

```java
public class QuizReview {
    public static void main(String[] args) {
        String s1 = "Java";
        String s2 = s1.concat(" rocks");
        String s3 = s1 + "!";
        System.out.println("s1 = " + s1 + " | s2 = " + s2 + " | s3 = " + s3);   // s1 unchanged

        Integer boxed = Integer.valueOf(41);   // explicit wrapper
        Integer auto = 1;                      // autoboxing
        int sum = boxed + auto;                // unboxing
        Double price = Double.valueOf(9.5);
        System.out.println("sum = " + sum + ", price * 2 = " + (price.doubleValue() * 2));
        System.out.println("\"apple\".compareTo(\"banana\") < 0? " + ("apple".compareTo("banana") < 0));
    }
}

QuizReview.main(null);
```
