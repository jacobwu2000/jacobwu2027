# Study Notes: Object-Oriented Polymorphism in Java

## 1. Core Definitions & Learning Targets

### What is Polymorphism?
Polymorphism literally means "many forms." In object-oriented programming, it allows objects of different classes to be treated as objects of a common superclass while executing behaviors specific to their actual subclass.

* **Polymorphic Reference Variable:** A reference variable that can store references to objects of its declared class or any subclass of its declared class.
* **Polymorphic Method:** A method overridden in at least one subclass, causing different behaviors depending on the actual object executing it.
* **Core Execution Principle:** Polymorphism is the act of executing an overridden non-static method from the correct class at **runtime** based on the actual object type, rather than the reference type.

---

## 2. Static vs. Dynamic Types

| Type Category | Definition | Example Context (`Shape s = new Triangle();`) |
| :--- | :--- | :--- |
| **Static Type** | Declared type of the reference variable at compile time. Determines which methods can be called. | `Shape` |
| **Dynamic Type** | Actual class of the object instantiated in memory at runtime. Determines which overridden method implementation executes. | `Triangle` |

### Method Resolution Chain
When calling `myObject.calc_area()`:
1. **Compile Time:** The compiler checks if `calc_area()` exists in the **static type** (`Shape`). If not, it throws a compilation error.
2. **Run Time:** The JVM checks the **dynamic type** (`Triangle`) for an overridden version of `calc_area()`.
3. If found, it executes the child version; otherwise, it moves up the class hierarchy to execute the parent implementation.

---

## 3. Compile-Time vs. Run-Time Mechanics

### Compile Time (Writing Code)
* Checks that methods exist on the **static type**.
* If a method only exists in a subclass, calling it directly via a superclass reference variable causes a **compile error**.

### Run Time (Executing Code)
* Evaluates the **dynamic type** to determine which overridden method code executes.
* Child-only methods are invisible at compile time unless **down-casting** is used.

### Down-Casting
Explicitly converting a superclass reference variable back to a subclass type to access subclass-specific functionality.

```java
Shape myShape = new Square(); 
// myShape.getSides(); // COMPILE ERROR: getSides() is not in Shape

Square mySquare = (Square) myShape; // Down-casting
System.out.println(mySquare.getSides()); // Success!

Water lake = new Lake(); 
System.out.println(lake.typeOfWater()); // Output: "lake"

int getSizePlusOne(Shape s) { 
    return s.getSize() + 1; 
}

Shape genericShape = new Shape();
Square square = new Square();

getSizePlusOne(genericShape); // Uses Shape.getSize()
getSizePlusOne(square);       // Uses Square.getSize() via polymorphism

Furniture myTable = new CoffeeTable();
Table myOtherTable = (Table) myTable;

// Both calls execute CoffeeTable's getName() method!
System.out.println(myTable.getName());      // Output: "Table Coffee"
System.out.println(myOtherTable.getName()); // Output: "Table Coffee"

Object banana = new Banana();

// Executes Banana's overridden toString() and equals() implementations
System.out.println(banana.toString()); 
System.out.println(banana.equals(null));

class Animal {
    public void makeSound() {
        System.out.println("Generic animal sound");
    }
}

class Dog extends Animal {
    @Override
    public void makeSound() {
        System.out.println("Bark");
    }

    public void fetch() {
        System.out.println("Fetching ball!");
    }
}

// Instantiate polymorphically
Animal myPet = new Dog();
myPet.makeSound(); // Outputs: "Bark"

// myPet.fetch(); // Compile error: Animal class lacks fetch()

Dog myDog = (Dog) myPet; // Down-cast Animal reference to Dog
myDog.fetch();           // Output: "Fetching ball!"
