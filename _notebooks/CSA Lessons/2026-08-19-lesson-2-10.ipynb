# Implementing String Algorithms

## Learning Objectives
When working with algorithms involving Java `String` objects, you should be able to:
* **Identify** standard String algorithms.
* **Modify** existing String algorithms to fit new specifications.
* **Develop** custom algorithms using String manipulation techniques.

---

## Key Java String Tools
* **`String.substring(start, end)`:** Retrieves a portion of a string from index `start` up to (but not including) `end`.
* **`String.equals(otherString)`:** Compares content equality between two string objects (**never** use `==`).
* **`String.length()`:** Returns the total number of characters in the string.
* **`for` Loops:** Used to iterate through individual characters or substring "windows" of a string.

---

## Substring Search: The "Window" Method

To find a target substring within a main string, create a moving "window" equal to the length of the substring and shift it through the main string index-by-index.
