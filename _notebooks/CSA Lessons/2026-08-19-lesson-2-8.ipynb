int sum = 0;
int sum_all = 0;

for (int i = 0; i < 10; i++) {
    if (i % 2 == 0) {
        sum += i;     // Adds even numbers only (0 + 2 + 4 + 6 + 8)
    }
    sum_all += i;     // Adds every number (0 through 9)
}

System.out.println(sum);     // Output: 20
System.out.println(sum_all); // Output: 45

# 4.2 for Loops

## for Loops vs. while Loops
* Both are iterative statements that repeatedly execute a code block as long as a condition is met.
* **while Loops:** Use when the total number of iterations is **unknown**.
* **for Loops:** Use when the exact number of iterations is **known**.

### Code Comparison

```java
// while loop approach
int i = 0;
while (i < 5) {
    System.out.println(i);
    i++;
}

// for loop approach
for (int i = 0; i < 5; i++) {
    System.out.println(i);
}
