# 1.14 Calling Instance Methods

Original lesson: [1.14 Calling Instance Methods](https://pages.opencodingsociety.com/csa/unit_01/1_14) · [homework page](https://pages.opencodingsociety.com/csa/unit_01/1_14_homework)

**Key idea:** an instance method runs on one specific object, using `object.method()`, and reads or changes that object's own fields. A static method belongs to the class and is called on the class name. Calling an instance method on a `null` reference throws a `NullPointerException`.

## Popcorn Hack #1: Dog with a breed

I added the field first, then the method, then created the object, in the order the lesson's tip suggests.

```java
// Work by Jacob Wu

public class Dog {
    private String name;
    private String breed;   // new property

    public Dog(String name, String breed) {
        this.name = name;
        this.breed = breed;
    }

    public void bark() {
        System.out.println(name + " says: Woof!");
    }

    // New instance method: uses this dog's own name and breed
    public void describe() {
        System.out.println("My dog " + name + " is a " + breed + "!");
    }
}

// To use it:
Dog d = new Dog("Fido", "Golden Retriever");
d.bark();
d.describe();

Dog second = new Dog("Biscuit", "Corgi");   // a second object keeps its own fields
second.describe();
```

## Popcorn Hack #2: Counter with five methods

Each new method follows the same pattern as `add`. It reads `count`, combines it with the parameter, and stores the result back in `count`.

```java
// Work by Jacob Wu

public class Counter {
    private int count;

    public void add(int x) {
        count += x;
    }

    public void subtract(int x) {
        count -= x;
    }

    public void multiply(int x) {
        count *= x;
    }

    // Integer division; reject 0 so the error message says what went wrong
    public void divide(int x) {
        if (x == 0) {
            throw new IllegalArgumentException("Cannot divide the counter by zero");
        }
        count /= x;
    }

    public int getCount() {
        return count;
    }
}

// Use it: all five methods, each called on the object c
Counter c = new Counter();
c.add(5);
System.out.println("After add(5):      " + c.getCount());
c.subtract(2);
System.out.println("After subtract(2): " + c.getCount());
c.multiply(10);
System.out.println("After multiply(10): " + c.getCount());
c.divide(4);
System.out.println("After divide(4):   " + c.getCount());
int val = c.getCount();
System.out.println("Final value from getCount(): " + val);
```

```java
public class Rectangle {
    private int length;
    private int width;

    public Rectangle(int l, int w) {
        length = l;
        width = w;
    }

    public int getArea() {
        return length * width;
    }

    public void scale(int factor) {
        length *= factor;
        width *= factor;
    }
}

public class Book {
    private String title;
    private int pages;

    public Book(String t, int p) {
        title = t;
        pages = p;
    }

    public String getTitle() {
        return title;
    }

    public int getPages() {
        return pages;
    }

    public void addPages(int additional) {
        pages += additional;
    }
}

Rectangle rect = new Rectangle(3, 4);
rect.scale(2);
System.out.println("Q2: " + rect.getArea());

Book novel = new Book("Java Basics", 200);
novel.addPages(50);
System.out.println("Q5: " + novel.getPages());
```

## Homework: Student Grade Tracker

**Requirements checklist**

- `Student` class with the instance variables `name`, `totalPoints`, `numAssignments`
- Constructor that sets the name and starts the totals at 0
- Void instance methods: `addGrade`, `printReport`
- Non-void instance methods: `getAverage` (returns `double`) and `getLetterGrade` (returns `String`)
- `main` creates two `Student` objects and shows that they work independently

**Bonus (extra credit):** the class also keeps each individual grade in a list. This adds `getHighestGrade`, `getLowestGrade`, `dropLowestGrade`, `compareTo`, `reset`, and error checking for negative or invalid grades.

```java
import java.util.ArrayList;

public class Student {
    // Required instance variables
    private String name;
    private int totalPoints;
    private int numAssignments;
    // Bonus: individual grades, needed for highest/lowest/drop-lowest
    private ArrayList<Integer> grades;

    // Sets the name and starts every total at zero
    public Student(String name) {
        this.name = name;
        this.totalPoints = 0;
        this.numAssignments = 0;
        this.grades = new ArrayList<>();
        System.out.println("Creating student: " + name);
        System.out.println("Student created successfully!");
    }

    // Void: records one grade and updates both running totals
    public void addGrade(int points) {
        if (points < 0 || points > 100) {
            throw new IllegalArgumentException("Grade must be 0-100, got " + points + " for " + name);
        }
        totalPoints += points;
        numAssignments++;
        grades.add(points);
        System.out.println("Adding grade: " + points + " points");
    }

    // Non-void: average rounded to 2 decimals; cast to double so it is not integer division
    public double getAverage() {
        if (numAssignments == 0) {
            return 0.0;
        }
        double average = (double) totalPoints / numAssignments;
        return Math.round(average * 100) / 100.0;
    }

    // Non-void: letter from the 90/80/70/60 scale. Checked from the top down,
    // so the first cutoff the average reaches decides the letter.
    public String getLetterGrade() {
        double average = getAverage();
        if (average >= 90) return "A";
        if (average >= 80) return "B";
        if (average >= 70) return "C";
        if (average >= 60) return "D";
        return "F";
    }

    // Helper used by printReport: an encouraging message for each letter
    private String getStatus() {
        switch (getLetterGrade()) {
            case "A": return "Excellent work!";
            case "B": return "Great job!";
            case "C": return "Keep working hard!";
            case "D": return "You can do it, ask for help!";
            default:  return "Let's make a plan to improve.";
        }
    }

    public String getFirstName() {
        return name.split(" ")[0];
    }

    // Void: prints this student's formatted report
    public void printReport() {
        System.out.println("--- " + getFirstName() + "'s Grade Report ---");
        System.out.println("Student Name: " + name);
        System.out.println("Total Points: " + totalPoints);
        System.out.println("Assignments Completed: " + numAssignments);
        System.out.println("Current Average: " + getAverage());
        System.out.println("Letter Grade: " + getLetterGrade());
        System.out.println("Status: " + getStatus());
    }

    public String getSummary() {
        return name + " - Average: " + getAverage() + " (" + getLetterGrade() + ")";
    }

    // ---- Bonus methods ----

    public int getHighestGrade() {
        int highest = grades.get(0);
        for (int grade : grades) {
            highest = Math.max(highest, grade);
        }
        return highest;
    }

    public int getLowestGrade() {
        int lowest = grades.get(0);
        for (int grade : grades) {
            lowest = Math.min(lowest, grade);
        }
        return lowest;
    }

    // Removes the lowest grade from the list and from both totals so the average updates
    public void dropLowestGrade() {
        int lowest = getLowestGrade();
        grades.remove(Integer.valueOf(lowest));
        totalPoints -= lowest;
        numAssignments--;
    }

    // Negative if this student's average is lower, 0 if equal, positive if higher
    public int compareTo(Student other) {
        return Double.compare(getAverage(), other.getAverage());
    }

    public void reset() {
        totalPoints = 0;
        numAssignments = 0;
        grades.clear();
    }

    public static void main(String[] args) {
        String divider = "\n========================================\n";
        System.out.println("=== Student Grade Tracker System ===\n");

        Student emma = new Student("Emma Rodriguez");
        System.out.println("\n--- Adding Grades for Emma ---");
        emma.addGrade(95);
        emma.addGrade(88);
        emma.addGrade(92);
        emma.addGrade(85);
        System.out.println();
        emma.printReport();
        System.out.println(divider);

        Student james = new Student("James Wilson");
        System.out.println("\n--- Adding Grades for James ---");
        james.addGrade(78);
        james.addGrade(82);
        james.addGrade(75);
        System.out.println();
        james.printReport();
        System.out.println(divider);

        System.out.println("Final Summary:");
        System.out.println(emma.getSummary());
        System.out.println(james.getSummary());

        // ---- Bonus demo ----
        System.out.println("\n--- Bonus Features ---");
        System.out.println("Emma highest: " + emma.getHighestGrade() + ", lowest: " + emma.getLowestGrade());
        System.out.println("Emma vs James compareTo: " + emma.compareTo(james) + " (positive = Emma is ahead)");
        james.dropLowestGrade();
        System.out.println("After dropping James's lowest grade: " + james.getSummary());
        try {
            emma.addGrade(-10);
        } catch (IllegalArgumentException error) {
            System.out.println("Rejected invalid grade: " + error.getMessage());
        }
        emma.reset();
        System.out.println("After reset: " + emma.getSummary());
    }
}

Student.main(null);
```

The output matches the sample: Emma's 90.0 is an A and James's 78.33 is a C. The two `Student` objects keep separate totals because each instance method call (`emma.addGrade(...)`, `james.addGrade(...)`) changes only the fields of the object it was called on.
