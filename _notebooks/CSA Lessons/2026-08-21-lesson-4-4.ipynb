# Array Traversals

## Traversing Array (1D)
* **Traversal** means visiting or accessing every single element in a data structure exactly once.
* Array traversal can be accomplished using **standard for loops**, **while loops**, or **enhanced for loops (for-each loops)**.

---

## Standard For Loop
An array in Java is indexed from `0` to `list.length - 1`.

### Review on For Loop Structure
```java
for (init; condition; incr/decr) {
    // statement(s)
}

import java.util.Random;

public class RandomArray {
    public static void main(String[] args) {
        int[] list = new int[6];
        Random rand = new Random();

        // FOR LOOP 1: Populate the array with random numbers (0 to 3)
        for (int i = 0; i < list.length; i++) {
            list[i] = rand.nextInt(4);
        }

        // FOR LOOP 2: Print each element using an enhanced for loop
        for (int element : list) {
            System.out.println(element);
        }
    }
}
