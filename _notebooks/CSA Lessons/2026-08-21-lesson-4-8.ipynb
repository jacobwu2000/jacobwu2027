## 7.1: ArrayList Intro

### Core Concepts
* **Dynamic vs. Static:** `ArrayList` objects are dynamic (their size grows or shrinks automatically), whereas standard Java arrays are static (fixed length).
* **Import Requirement:** `ArrayList` is part of `java.util`, so it must be imported:
  ```java
  import java.util.ArrayList;

import java.util.ArrayList;

public class ArrayListMethodsExample {
    private String manipulateList() {
        ArrayList<String> list = new ArrayList<>();
        
        // 1. Add 2 items
        list.add("Item 1");
        list.add("Item 2");
        
        // Save initial state for report
        String addedItems = list.get(0) + ", " + list.get(1);
        
        // 2. Add extra items for demonstration
        list.add("Item 3");
        list.add("Item 4");
        
        // Remove an item of choice (e.g., index 1)
        String removed = list.remove(1);
        
        // 3. Replace an item of choice (e.g., index 0)
        String replaced = list.set(0, "Replaced Item");
        
        // 4. Get first and last element dynamically
        String first = list.get(0);
        String last = list.get(list.size() - 1);
        
        // 5. Return summary in one string
        return "Added: [" + addedItems + "], Removed: " + removed + 
               ", Replaced: " + replaced + ", First: " + first + 
               ", Last: " + last + ", Final Size: " + list.size();
    }

    public static void main(String[] args) {
        ArrayListMethodsExample example = new ArrayListMethodsExample();
        String output = example.manipulateList();
        System.out.println(output);
    }
}

import java.util.ArrayList;

public class Main {
    public static void main(String[] args) {
        ArrayList<String> roster = new ArrayList<>();
        roster.add("Hello");
        roster.add("World");
        roster.add("Java");

        int sum = 0;
        for (int i = 0; i < roster.size(); i++) {
            String element = roster.get(i);
            if (element != null) {
                sum += element.length();
            }
        }
        System.out.println("Total Length Sum: " + sum); // Output: 14
    }
}
